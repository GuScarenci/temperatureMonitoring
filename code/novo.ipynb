```python
import requests
import json

BASE_URL = "http://andromeda.lasdpc.icmc.usp.br:60106"

documentacao = requests.get(
    f"{BASE_URL}/v3/api-docs",
    timeout=15
).json()

endpoints_interessantes = [
    "/guest/filter",
    "/guest/latest",
    "/datastream/filter",
    "/guest/environment",
    "/guest/deviceTypes",
    "/device"
]

for endpoint in endpoints_interessantes:
    print(f"\n{'=' * 70}")
    print(endpoint)
    print("=" * 70)

    if endpoint not in documentacao["paths"]:
        print("Endpoint não encontrado.")
        continue

    for metodo, detalhes in documentacao["paths"][endpoint].items():
        print(f"\nMétodo: {metodo.upper()}")
        print("Descrição:", detalhes.get("description", ""))
        print("Resumo:", detalhes.get("summary", ""))

        if "requestBody" in detalhes:
            print("\nREQUEST BODY:")
            print(json.dumps(
                detalhes["requestBody"],
                indent=2,
                ensure_ascii=False
            ))

        print("\nRESPONSES:")
        print(json.dumps(
            detalhes.get("responses", {}),
            indent=2,
            ensure_ascii=False
        ))
```

```python
schemas = documentacao["components"]["schemas"]

schemas_interessantes = [
    nome for nome in schemas
    if any(
        termo in nome.lower()
        for termo in [
            "guest",
            "stream",
            "filter",
            "device",
            "environment"
        ]
    )
]

for nome in schemas_interessantes:
    print(f"\n{'=' * 70}")
    print(nome)
    print("=" * 70)

    print(json.dumps(
        schemas[nome],
        indent=2,
        ensure_ascii=False
    ))
```

```python
import os
from getpass import getpass

import requests
from getpass import getpass

BASE_URL = "http://andromeda.lasdpc.icmc.usp.br:60106"

session = requests.Session()

username = os.environ.get("SMARTLAB_USERNAME") or input("Usuário Smartlab: ")
password = os.environ.get("SMARTLAB_PASSWORD") or getpass("Senha Smartlab: ")

response = session.post(
    f"{BASE_URL}/auth/login",
    json={
        "username": username,
        "password": password
    },
    timeout=15
)

del password
response.raise_for_status()
print("Status:", response.status_code)

if response.ok:
    print("Autenticação aceita.")
    print("Cookies de sessão:", list(session.cookies.keys()))
    print("Campos da resposta:",
          list(response.json().keys()) if response.headers.get(
              "Content-Type", ""
          ).startswith("application/json") else "Resposta não JSON")
else:
    print("Falha na autenticação:", response.status_code)
```

```python
token = response.json()["token"]

session.headers.update({
    "Authorization": f"Bearer {token}"
})

print("Token configurado na sessão.")
```

```python
def consultar_temperatura(
    id_device,
    inicio,
    fim,
    time_bucket="5 minutes"
):
    payload = {
        "startDate": inicio,
        "endDate": fim,
        "idDeviceType": 1,
        "idEnviroment": 3,
        "idDevice": id_device,
        "timeBucket": time_bucket
    }

    resposta = session.post(
        f"{BASE_URL}/datastream/filter",
        json=payload,
        timeout=60
    )

    resposta.raise_for_status()

    dados = resposta.json()

    df = pd.DataFrame(dados)

    if df.empty:
        return pd.DataFrame(columns=[id_device])

    df["time_interval"] = pd.to_datetime(
        df["time_interval"],
        utc=True
    )

    df["avg_value"] = pd.to_numeric(
        df["avg_value"],
        errors="coerce"
    )

    df = df.set_index("time_interval")

    return df[["avg_value"]].rename(
        columns={"avg_value": id_device}
    )
```
